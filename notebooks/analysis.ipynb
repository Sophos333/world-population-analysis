```python
# Load dataset
df = pd.read_csv("../data/world_population.csv")

# Display first 5 rows
df.head()

# Check for missing values
df.isnull().sum()
df.columns
df.info()
df.describe()
df.rename(columns={
    "2022 Population": "Population_2022",
    "2020 Population": "Population_2020",
    "2015 Population": "Population_2015",
    "2010 Population": "Population_2010",
    "2000 Population": "Population_2000",
    "1990 Population": "Population_1990",
}, inplace=True)
df.sort_values(by="Population_2022", ascending=False).head(10)
df["Growth_1990_2022"] = df["Population_2022"] - df["Population_1990"]
df[["Country/Territory", "Growth_1990_2022"]].sort_values(by="Growth_1990_2022", ascending=False).head(10)
df.groupby("Continent")["Population_2022"].sum().sort_values(ascending=False)
```

```python
# Calculate population growth from 1990 to 2022
df["Growth_1990_2022"] = df["Population_2022"] - df["Population_1990"]

# Top 10 countries with highest population growth
top_growth = df.sort_values(by="Growth_1990_2022", ascending=False).head(10)

# Plot growth
plt.figure(figsize=(10,6))
plt.barh(top_growth["Country/Territory"], top_growth["Growth_1990_2022"])
plt.xlabel("Population Growth (1990 to 2022)")
plt.title("Top 10 Countries by Population Growth (1990–2022)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
